# Load Dataset Notebook

Run the cell below to download the dataset and save it to the `data` folder. All data is in csv format. 

Within data, there will be subfolders naming regions, and within those subfolders, there will be csv files for each region.

```python
# Import HF token from env file
import os
from dotenv import load_dotenv
from datasets import load_dataset

# Load the .env file
load_dotenv()

# Get the HF token from the environment variable
hf_token = os.getenv("HF_API_KEY")

# Load your private dataset
dataset = load_dataset(
    "shaheerdeeplearning2/SentinelDataset", 
    token=hf_token
)

# Save the dataset to the `data` folder
dataset.save_to_disk("data")
```
